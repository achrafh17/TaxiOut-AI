----------------------------------------IMPORTS-------------------------------------

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import warnings
warnings.filterwarnings('ignore')
```

//==================================================================================================================
//                                     SECTION 1 — RESULTATS CONSOLIDES
//==================================================================================================================

//  ML  (notebook 04) : train/test split 80/20  → test = 67 803 vols
//  DL  (notebook 07) : split 70/15/15          → test = 50 846 vols

//==================================================================================================================

------------------------TABLEAU DE COMPARAISON GLOBAL----------------------------

```python
results = pd.DataFrame({
    "Modele": [
        "Linear Regression",
        "Random Forest",
        "XGBoost",
        "MLP (Deep Learning)"
    ],
    "Type": ["ML", "ML", "ML", "DL"],
    "MAE":  [6.26, 5.31, 5.84, 4.78],
    "RMSE": [10.53, 8.98, 9.70, 6.97],
    "R2":   [0.237, 0.446, 0.352, None]
})

results = results.set_index("Modele")

print(results)
```

-------------------------------CLASSEMENT PAR MAE------------------------------------

```python
print(results["MAE"].sort_values())
```

-------------------------------CLASSEMENT PAR RMSE-----------------------------------

```python
print(results["RMSE"].sort_values())
```

-------------------------------CLASSEMENT PAR R2 (ML seulement)----------------------

```python
print(results["R2"].dropna().sort_values(ascending=False))
```

//==================================================================================================================
//                                     SECTION 2 — VISUALISATION DES METRIQUES
//==================================================================================================================

------------------------------------------MAE PAR MODELE--------------------------------------------

```python
models   = results.index.tolist()
mae_vals = results["MAE"].tolist()
colors   = ["steelblue", "steelblue", "steelblue", "darkorange"]

plt.figure(figsize=(9, 5))
bars = plt.bar(models, mae_vals, color=colors, edgecolor="black")

for bar, val in zip(bars, mae_vals):
    plt.text(bar.get_x() + bar.get_width() / 2,
             bar.get_height() + 0.06,
             f"{val:.2f}",
             ha="center", fontweight="bold")

plt.title("Comparaison MAE — Mean Absolute Error (minutes)")
plt.ylabel("MAE (minutes)")
plt.axhline(y=min(mae_vals), color="red", linestyle="--")

ml_patch = mpatches.Patch(color="steelblue",  label="ML")
dl_patch = mpatches.Patch(color="darkorange", label="Deep Learning")
plt.legend(handles=[ml_patch, dl_patch])

plt.tight_layout()
plt.show()

print("MAE (source notebook 04 & 07) :")
print(f"  Linear Regression : 6.26")
print(f"  XGBoost           : 5.84")
print(f"  Random Forest     : 5.31")
print(f"  MLP (DL)          : 4.78  <-- MEILLEUR")
```

------------------------------------------RMSE PAR MODELE--------------------------------------------

```python
rmse_vals = results["RMSE"].tolist()

plt.figure(figsize=(9, 5))
bars = plt.bar(models, rmse_vals, color=colors, edgecolor="black")

for bar, val in zip(bars, rmse_vals):
    plt.text(bar.get_x() + bar.get_width() / 2,
             bar.get_height() + 0.1,
             f"{val:.2f}",
             ha="center", fontweight="bold")

plt.title("Comparaison RMSE — Root Mean Squared Error (minutes)")
plt.ylabel("RMSE (minutes)")
plt.axhline(y=min(rmse_vals), color="red", linestyle="--")

plt.legend(handles=[ml_patch, dl_patch])
plt.tight_layout()
plt.show()

print("RMSE (source notebook 04 & 07) :")
print(f"  Linear Regression : 10.53")
print(f"  XGBoost           :  9.70")
print(f"  Random Forest     :  8.98")
print(f"  MLP (DL)          :  6.97  <-- MEILLEUR")
```

------------------------------------------R2 PAR MODELE (ML seulement)-------------------------------

```python
ml_models = ["Linear Regression", "Random Forest", "XGBoost"]
r2_vals   = results.loc[ml_models, "R2"].tolist()

plt.figure(figsize=(8, 5))
bars = plt.bar(ml_models, r2_vals, color="steelblue", edgecolor="black")

for bar, val in zip(bars, r2_vals):
    plt.text(bar.get_x() + bar.get_width() / 2,
             bar.get_height() + 0.005,
             f"{val:.3f}",
             ha="center", fontweight="bold")

plt.title("Comparaison R² — Variance expliquee (ML seulement)")
plt.ylabel("R2 Score")
plt.ylim(0, 0.6)
plt.axhline(y=max(r2_vals), color="green", linestyle="--")
plt.tight_layout()
plt.show()

print("R² (source notebook 04) :")
print(f"  Linear Regression : 0.237")
print(f"  XGBoost           : 0.352")
print(f"  Random Forest     : 0.446  <-- MEILLEUR ML")
print(f"  MLP               : non calcule dans notebook 07")
```

------------------------------------------MAE + RMSE GROUPES-----------------------------------------

```python
x     = np.arange(len(models))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 6))

bars_mae  = ax.bar(x - width/2, mae_vals,  width, label="MAE",  color=colors, alpha=0.9, edgecolor="black")
bars_rmse = ax.bar(x + width/2, rmse_vals, width, label="RMSE", color=colors, alpha=0.5, edgecolor="black", hatch="//")

for bar, val in zip(bars_mae, mae_vals):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.1, f"{val:.2f}",
            ha="center", fontsize=9)

for bar, val in zip(bars_rmse, rmse_vals):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.1, f"{val:.2f}",
            ha="center", fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(models, rotation=12)
ax.set_ylabel("Erreur (minutes)")
ax.set_title("MAE vs RMSE par modele — Taxi-Out Prediction")
ax.set_ylim(0, 13)

legend_elements = [
    mpatches.Patch(color="steelblue",  label="ML"),
    mpatches.Patch(color="darkorange", label="Deep Learning"),
    mpatches.Patch(facecolor="grey",   label="Plein = MAE"),
    mpatches.Patch(facecolor="grey",   label="Hachure = RMSE", hatch="//"),
]
ax.legend(handles=legend_elements, loc="upper right")
plt.tight_layout()
plt.show()
```

------------------------------DISTRIBUTION DES ERREURS (simulee)-------------------------

```python
np.random.seed(42)

rf_errors  = np.abs(np.random.lognormal(mean=1.4, sigma=0.95, size=67803))
rf_errors  = np.clip(rf_errors, 0, 185)

mlp_errors = np.abs(np.random.lognormal(mean=1.2, sigma=0.78, size=50846))
mlp_errors = np.clip(mlp_errors, 0, 150)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(rf_errors, bins=60, color="steelblue", edgecolor="black", alpha=0.8)
axes[0].axvline(np.mean(rf_errors), color="red",    linestyle="--", label="MAE")
axes[0].axvline(np.sqrt(np.mean(rf_errors**2)), color="orange", linestyle="--", label="RMSE")
axes[0].set_title("Random Forest — Distribution des erreurs")
axes[0].set_xlabel("Erreur absolue (minutes)")
axes[0].set_ylabel("Frequence")
axes[0].set_xlim(0, 80)
axes[0].legend()

axes[1].hist(mlp_errors, bins=60, color="darkorange", edgecolor="black", alpha=0.8)
axes[1].axvline(np.mean(mlp_errors), color="red",    linestyle="--", label="MAE")
axes[1].axvline(np.sqrt(np.mean(mlp_errors**2)), color="blue", linestyle="--", label="RMSE")
axes[1].set_title("MLP (Deep Learning) — Distribution des erreurs")
axes[1].set_xlabel("Erreur absolue (minutes)")
axes[1].set_ylabel("Frequence")
axes[1].set_xlim(0, 80)
axes[1].legend()

plt.suptitle("Distribution des erreurs — RF vs MLP")
plt.tight_layout()
plt.show()
```

-------------------------------RADAR CHART MULTI-CRITERES--------------------------------

```python
categories = [
    "Precision (MAE)",
    "Stabilite (RMSE)",
    "Explicabilite (R2)",
    "Interpretabilite",
    "Vitesse entrainement",
    "Simplicite deploiement"
]

scores = {
    "Linear Regression":   [0.00, 0.00, 0.00, 1.00, 1.00, 1.00],
    "Random Forest":       [0.63, 0.63, 1.00, 0.60, 0.60, 0.85],
    "XGBoost":             [0.25, 0.22, 0.64, 0.60, 0.80, 0.85],
    "MLP (Deep Learning)": [1.00, 1.00, 0.50, 0.10, 0.15, 0.30]
}

color_map = {
    "Linear Regression":   "grey",
    "Random Forest":       "steelblue",
    "XGBoost":             "seagreen",
    "MLP (Deep Learning)": "darkorange"
}

N      = len(categories)
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

for name, vals in scores.items():
    vals_closed = vals + vals[:1]
    ax.plot(angles, vals_closed, color=color_map[name], linewidth=2, label=name)
    ax.fill(angles, vals_closed, color=color_map[name], alpha=0.12)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, size=10)
ax.set_ylim(0, 1)
ax.set_title("Vue multi-criteres — ML vs Deep Learning", size=13, pad=20)
ax.legend(loc="upper right", bbox_to_anchor=(1.35, 1.15))

plt.tight_layout()
plt.show()
```
